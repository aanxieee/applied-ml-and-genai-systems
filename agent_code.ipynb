```python
from diffusers import StableDiffusionXLPipeline, DDPMScheduler, UNet2DConditionModel
from diffusers.utils import make_image_grid
from peft import get_peft_model, LoraConfig
from transformers import CLIPTokenizer
import torch
from torch.utils.data import Dataset, DataLoader
from PIL import Image
import os

# Paths
image_dir = "/content/adimagenet_images"
output_dir = "/content/adora_ssd1b_lora"

# Load pretrained model
pipe = StableDiffusionXLPipeline.from_pretrained(
    "segmind/SSD-1B",
    torch_dtype=torch.float16,
    variant="fp16"
).to("cuda")

tokenizer = pipe.tokenizer

# Custom Dataset
class AdoraDataset(Dataset):
    def __init__(self, image_folder):
        self.image_folder = image_folder
        self.images = sorted([f for f in os.listdir(image_folder) if f.endswith(".png")])

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        image_path = os.path.join(self.image_folder, self.images[idx])
        caption_path = image_path.replace(".png", ".txt")

        image = Image.open(image_path).convert("RGB").resize((512, 512))
        caption = open(caption_path).read()
        pixel_values = pipe.feature_extractor(images=image, return_tensors="pt").pixel_values[0]

        inputs = tokenizer(caption, return_tensors="pt", padding="max_length", truncation=True, max_length=77)
        return {
            "pixel_values": pixel_values,
            "input_ids": inputs.input_ids[0]
        }

# Dataloader
train_dataset = AdoraDataset(image_dir)
train_loader = DataLoader(train_dataset, batch_size=1, shuffle=True)

# LoRA Config
lora_config = LoraConfig(
    r=4,
    lora_alpha=16,
    target_modules=["transformer_blocks"],  # Adjust this based on model architecture
    lora_dropout=0.1,
    bias="none",
    task_type="CAUSAL_LM"
)

# Inject LoRA
pipe.unet = get_peft_model(pipe.unet, lora_config)

# Optimizer
optimizer = torch.optim.Adam(pipe.unet.parameters(), lr=1e-5)

# Training Loop (Very Basic)
pipe.unet.train()
for epoch in range(1):
    for step, batch in enumerate(train_loader):
        inputs = batch["input_ids"].unsqueeze(0).to("cuda")
        pixel_values = batch["pixel_values"].unsqueeze(0).to("cuda")
        loss = pipe.unet(pixel_values, inputs).loss
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()
        if step % 10 == 0:
            print(f"Epoch {epoch} Step {step} - Loss: {loss.item():.4f}")
```
