```python
#  Archie Agent: Rule-Based Clause Detection
### This project identifies key legal clauses in contract documents using keyword-based rules. It is designed to help quickly audit contracts for required legal information.
```

```python
# --- Step 1: Install + Upload Files ---
from google.colab import files
uploaded = files.upload()  # This allows file upload in Colab
```

```python

# --- Step 2: Imports ---
import os
import re
import pandas as pd
```

```python
# --- Step 3: Clause Definitions ---
CLAUSES = {
    "Termination Clause": ["termination clause", "termination notice", "contract termination"],
    "Payment Terms": ["payment terms", "invoice due", "billing", "collection costs"],
    "Governing Law": ["governing law", "jurisdiction", "legal disputes", "state of"],
    "Service Terms": ["service level agreement", "sla", "support terms", "uptime"],
    "Pricing Info": ["cost breakdown", "total price", "pricing", "charges", "projected cost"],
}
```

```python
# --- Step 4: Text Cleaning Function ---
def clean_text(text):
    text = re.sub(r'\s+', ' ', text)
    text = re.sub(r'[^\w\s.,:$%-]', '', text)
    return text.lower().strip()
```

```python

# --- Step 5: Clause Detection Functions ---
def detect_clause(text, keywords):
    return any(keyword in text for keyword in keywords)
```

```python
def evaluate(text):
    return {
        clause: ("Found" if detect_clause(text, kws) else "Missing")
        for clause, kws in CLAUSES.items()
    }
```

```python
# --- Step 6: Print Results Per File ---
def process_files(file_list):
    for file in file_list:
        if file.endswith('.txt'):
            print(f"\n📄 File: {file}")
            with open(file, 'r', encoding='utf-8') as f:
                raw = f.read()
            cleaned = clean_text(raw)
            results = evaluate(cleaned)
            for clause, result in results.items():
                icon = "yes" if result == "Found" else "No"
                print(f"{clause}: {icon} {result}")
```

```python
# --- Step 7: Create Summary Table for All Files ---
def create_summary_table(file_list):
    summary_data = []
    for file in file_list:
        if os.path.exists(file):
            with open(file, 'r', encoding='utf-8') as f:
                raw = f.read()
            cleaned = clean_text(raw)
            result = evaluate(cleaned)
            result['File'] = file
            summary_data.append(result)
        else:
            print(f" File not found: {file}")
    df = pd.DataFrame(summary_data)
    return df.set_index("File")
```

```python

# --- Step 8: Extract Uploaded Files List ---
file_list = list(uploaded.keys())
```

```python
# --- Step 9: Run Detection ---
process_files(file_list)
```

```python
# --- Step 10: Generate Summary Table and Export CSV ---
df_summary = create_summary_table(file_list)
print("\n Summary Table:")
print(df_summary)
```

```python
# --- Step 11: Save and Offer CSV for Download ---
df_summary.to_csv("clause_detection_summary.csv")
files.download("clause_detection_summary.csv")
print("\n Report saved and download started: 'clause_detection_summary.csv'")
```
